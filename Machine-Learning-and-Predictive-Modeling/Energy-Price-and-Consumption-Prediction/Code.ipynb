# Predicting the Price of Energy in Germany

```python
import pandas as pd 
import numpy as np 
import datetime
```

```python
# Data Ingestion 

df = pd.read_csv(r"")
df.head()
```

```python
# Feature Engineering

df["date"] = pd.to_datetime(df["date"])
df["year"] = df["date"].dt.year
df["month"] = df["date"].dt.month
df["day"] = df["date"].dt.day
df["dayofweek"] = df["date"].dt.dayofweek 
df["is_weekend"] = (df["dayofweek"] >= 5).astype(int)

df["month_sin"] = np.sin(2*np.pi*df["month"]/12)
df["month_cos"] = np.cos(2*np.pi*df["month"]/12)

df["price_(t-1)"] = df["energy_price"].shift(1)
df["price_ma7"] = df["energy_price"].rolling(window=7).mean()
df["price_std7"] = df["energy_price"].rolling(window=7).std()



df.head()
```

```python
# Data Cleaning 

df = df.dropna()
df = df.drop_duplicates()
df.head()
```
